```python
# Enhanced Three-Pathway Arrhythmogenesis Flowchart
# With additional mechanistic details from the text

arrhythmogenesis_flowchart_detailed = {
    "Epicardial Adipose Tissue": {
        "Characteristics": {
            "Location": "Between myocardium and visceral pericardium",
            "Barrier": "Absent fascial barrier - direct paracrine exchange",
            "Cellular origin": "90% of adipokines from nonadipocyte stromal-vascular fraction",
            "Correlations": "Higher thickness → systemic inflammation, cardiac fibrosis"
        },
        "Pathways": {
            "Electrophysiological": {
                "Key Mediators": ["IL-6", "IL-1β", "FABP4"],
                "Mechanisms": [
                    "IL-6: Modulates L-type Ca²⁺ channels",
                    "IL-6: Suppresses SERCA2 expression/function",
                    "IL-1β: Enhances spontaneous SR Ca²⁺ release",
                    "IL-1β: Reduces Ca²⁺ regulatory proteins",
                    "FABP4: Reduces systolic intracellular Ca²⁺"
                ],
                "Cellular Effects": "Disrupted calcium homeostasis",
                "Electrophysiological Consequences": "Altered AP duration, EADs, DADs",
                "Arrhythmia Substrate": "Triggered activity + re-entrant arrhythmias"
            },
            "Fibrotic": {
                "Key Mediators": ["Adipocytokines", "Inflammatory mediators", "Adjacent fatty deposits"],
                "Mechanisms": [
                    "Direct induction of fibrotic remodeling",
                    "Paracrine action on atrial myocardium",
                    "Immune/inflammatory-mediated fibrotic remodeling"
                ],
                "Cellular Effects": "Fibroblast activation, collagen deposition",
                "Electrophysiological Consequences": "Conduction barriers, slowed conduction",
                "Arrhythmia Substrate": "Micro-reentry circuits favoring AF"
            },
            "Structural/Inflammatory": {
                "Key Mediators": ["Chemokines", "Growth factors", "Remodeling factors"],
                "Mechanisms": [
                    "Proinflammatory cytokine profile",
                    "Transport of systemic inflammation to myocardium",
                    "Oxidative stress induction"
                ],
                "Cellular Effects": "Myocyte hypertrophy, matrix remodeling",
                "Electrophysiological Consequences": "Atrial enlargement, altered wall stress",
                "Arrhythmia Substrate": "Atrial hypertrophy, dilatation, heightened AF risk"
            }
        },
        "Clinical Correlations": {
            "Obesity": "Higher EAT thickness → systemic inflammation",
            "Heart Failure": "EAT thickness correlates with cardiac fibrosis",
            "AF Risk": "Cardiac fat releases promote atrial remodeling"
        }
    }
}

def print_detailed_flowchart(flowchart):
    for central, data in flowchart.items():
        print("="*80)
        print(f"CENTRAL NODE: {central}".center(80))
        print("="*80)

        # Print characteristics
        print("\n📌 CHARACTERISTICS:")
        for char, value in data["Characteristics"].items():
            print(f"  • {char}: {value}")

        # Print pathways
        print("\n🔬 THREE PATHWAYS TO ARRHYTHMOGENESIS:")
        print("-"*80)

        for pathway, details in data["Pathways"].items():
            print(f"\n▶ PATHWAY {pathway.upper()}")
            print(f"  Key Mediators: {', '.join(details['Key Mediators'])}")
            print(f"  Mechanisms:")
            for mech in details['Mechanisms']:
                print(f"    - {mech}")
            print(f"  Cellular Effects: {details['Cellular Effects']}")
            print(f"  Electrophysiological: {details['Electrophysiological Consequences']}")
            print(f"  Arrhythmia Substrate: {details['Arrhythmia Substrate']}")

        # Print clinical correlations
        print("\n🏥 CLINICAL CORRELATIONS:")
        for condition, correlation in data["Clinical Correlations"].items():
            print(f"  • {condition}: {correlation}")

        print("\n" + "="*80)

print_detailed_flowchart(arrhythmogenesis_flowchart_detailed)

# Enhanced visualization with graphviz
try:
    from graphviz import Digraph

    # Create a more sophisticated flowchart
    dot = Digraph(comment='EAT-Mediated Arrhythmogenesis',
                  format='png',
                  graph_attr={'rankdir': 'TB', 'splines': 'ortho'})

    # Style definitions
    dot.attr('node', shape='box', style='filled', fontname='Arial')

    # Central node
    dot.node('EAT', 'Epicardial Adipose Tissue\n(Secretory Organ)',
             fillcolor='lightblue', fontsize='12')

    # Pathway nodes
    pathways = {
        'Electrophysiological': {'color': 'lightcoral', 'label': 'ELECTROPHYSIOLOGICAL\n↓Ca²⁺ Handling\n↓SERCA2\n↑Triggered Activity'},
        'Fibrotic': {'color': 'lightgreen', 'label': 'FIBROTIC\nMicro-reentry\nAF Substrate'},
        'Structural': {'color': 'lightyellow', 'label': 'STRUCTURAL/INFLAMMATORY\nHypertrophy\nDilatation'}
    }

    for path, attrs in pathways.items():
        dot.node(path, attrs['label'], fillcolor=attrs['color'])
        dot.edge('EAT', path)

    # Add mechanistic details as subnodes
    dot.node('IL6', 'IL-6, IL-1β, FABP4', shape='ellipse', fillcolor='lightgray')
    dot.node('fibrosis', 'Adipocytokines\nInflammatory mediators', shape='ellipse', fillcolor='lightgray')
    dot.node('inflammation', 'Chemokines\nGrowth factors', shape='ellipse', fillcolor='lightgray')

    dot.edge('Electrophysiological', 'IL6')
    dot.edge('Fibrotic', 'fibrosis')
    dot.edge('Structural', 'inflammation')

    # Final consequence
    dot.node('AF', 'ATRIAL FIBRILLATION',
             shape='doublecircle', fillcolor='red', fontcolor='white', fontsize='14')
    dot.edge('Electrophysiological', 'AF')
    dot.edge('Fibrotic', 'AF')
    dot.edge('Structural', 'AF')

    # Render
    output_file = 'enhanced_arrhythmogenesis_flowchart'
    dot.render(output_file, cleanup=True)
    print(f"\n✓ Enhanced flowchart visualization saved as {output_file}.png")

except ImportError:
    print("\n⚠ Install graphviz for enhanced visualization: pip install graphviz")
```
